```python
# Conectar a Base SQLite
import sqlite3
conexion = sqlite3.connect('data/productos.db')
```

```python
# Crear tabla de productos dentro de la base de datos con la siguiente estructura:

create_sql = """CREATE TABLE productos (
id_producto VARCHAR(10) PRIMARY KEY,
nombre VARCHAR(100),
categoria VARCHAR(50),
marca VARCHAR(50),
precio_usd DECIMAL(10,2),
stock INTEGER,
color VARCHAR(50),
potencia_w INTEGER,
capacidad VARCHAR(20),
voltaje VARCHAR(20),
peso_kg DECIMAL(5,2),
garantia_meses INTEGER,
descripcion TEXT);"""

print(repr(create_sql))  # Shows hidden characters

cursor = conexion.cursor()
cursor.execute(create_sql)
conexion.commit()
```

```python
# Abrir archivo csv y cargar datos a la base de datos
# Lista de campos del csv: ID_Producto	Nombre	Categoria	Marca	Precio_USD	Stock	Color	Potencia_W	Capacidad	Voltaje	Peso_KG	Garantia_Meses	Descripcion

import csv
with open('data/productos.csv', 'r') as archivo_csv:
    lector_csv = csv.reader(archivo_csv)
    next(lector_csv)  # Saltar la primera fila (encabezados)
    for fila in lector_csv:
        conexion.execute('INSERT INTO productos (id_producto, nombre, categoria, marca, precio_usd, stock, color, potencia_w, capacidad, voltaje, peso_kg, garantia_meses, descripcion) VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)', fila)

# Confirmar los cambios y cerrar la conexión
conexion.commit()
```

```python
# Mostrar contenido de la tabla productos
cursor.execute('SELECT * FROM productos')
filas = cursor.fetchall()
for fila in filas:
    print(fila)
```
